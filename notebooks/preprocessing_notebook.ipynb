```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Load the raw dataset
# (Make sure the path matches your notebook location relative to the raw data)
df = pd.read_csv('../data/raw/diabetic_data.csv')

# 2. Data Cleaning
# In this dataset, missing values are usually represented by '?'
df.replace('?', np.nan, inplace=True)

# Drop columns with excessively high missing values (e.g., 'weight' is ~97% missing)
df.drop(columns=['weight', 'payer_code', 'medical_specialty'], inplace=True, errors='ignore')

# Drop rows where critical diagnostic data or demographics are missing
df.dropna(subset=['diag_1', 'diag_2', 'diag_3', 'race', 'gender'], inplace=True)

# 3. Feature Engineering & Synthetic Data Generation
np.random.seed(42) # For reproducibility
n_rows = len(df)

# Generate doctor_id
df['doctor_id'] = ['DOC_' + str(np.random.randint(1000, 9999)) for _ in range(n_rows)]

# Generate department
departments = ['Cardiology', 'Endocrinology', 'General Medicine', 'Neurology', 'Surgery']
df['department'] = np.random.choice(departments, n_rows)

# Generate admission_date (Random dates between 1999 and 2008)
start_date = datetime(1999, 1, 1)
end_date = datetime(2008, 12, 31)
date_range = (end_date - start_date).days
df['admission_date'] = [start_date + timedelta(days=np.random.randint(0, date_range)) for _ in range(n_rows)]

# Generate bed_occupancy (Percentage full at the time of admission)
df['bed_occupancy'] = np.random.randint(50, 100, n_rows)

# Generate location
locations = ['New York, NY', 'Los Angeles, CA', 'Chicago, IL', 'Houston, TX', 'Phoenix, AZ']
df['location'] = np.random.choice(locations, n_rows)

# Generate patient_satisfaction (1 to 5 scale)
df['patient_satisfaction'] = np.random.randint(1, 6, n_rows)

# length_of_stay (Mapping existing 'time_in_hospital' for clarity)
df['length_of_stay'] = df['time_in_hospital']

# 4. Save Cleaned Dataset
df.to_csv('../data/cleaned/cleaned_dataset.csv', index=False)

# 5. Data Encoding (For Machine Learning Squads)
# Creating a copy for the encoded version
df_encoded = df.copy()
categorical_cols = ['race', 'gender', 'department', 'location', 'readmitted', 'diabetesMed', 'change']

# One-Hot Encoding for categorical variables
df_encoded = pd.get_dummies(df_encoded, columns=categorical_cols, drop_first=True)

# Save Encoded Dataset
df_encoded.to_csv('../data/cleaned/encoded_dataset.csv', index=False)

# You can also save a 'final_dataset.csv' if there's a specific merged version required
df_encoded.to_csv('../data/cleaned/final_dataset.csv', index=False)

print("Data Preprocessing Complete!")
print(f"Cleaned Data Shape: {df.shape}")
print(f"Encoded Data Shape: {df_encoded.shape}")
```
